```python
# Q2 this is just practice with classes. I haven't worked with these before doing the datacamp practice
# and I wanted to use this time to get some reps in 

class Employee: 
    def __init__(self, name, pay = 0, hours = 0 ): 
        self.name = name 
        self.pay = pay 
        self.hours = hours

    def __str__(self): 
        return f"{self.name} makes ${self.pay} an hour and worked { self.hours} for a total of ${self.total_pay()} this week"
        
    def total_pay(self): 
        if (self.hours <= 20): 
            return self.hours * self.pay
        else:
            return (20 * self.pay) + ((self.hours - 20) * (self.pay * 1.5) )

emp = Employee("Buddy Budderson") 

txt = input("How many hours did Buddy work?: ")

txt1 = input("How much does Buddy get paid per hour?: ") 

emp.hours = int(txt)
emp.pay = int(txt1)

print(emp)
            
```

```python
name = "Alice"
age = 30
price = 19.995


# Positional arguments
print("Hello, {}. You are {}.".format(name, age))

# Indexed arguments
print("Hello, {0}. You are {1}. Repeat: {0}.".format(name, age))

# Keyword arguments
print("Price: {p:.2f}".format(p=price))
```

```python
class BankAccount: 
    def __init__(self, bankname, firstname, lastname, balance=0): 
        self.bankname = bankname 
        self.firstname = firstname
        self.lastname = lastname 
        self.balance = balance 
    def deposit (self, amount):
        self.balance += amount 
    def withdrawl (self, amount):
        if (amount <= self.balance ): 
            self.balance -= amount 
        else: 
            print("Error, not enough funds in account")
    def __str__(self): 
        return f"The bank name is {self.bankname}, the account holder is {self.firstname} {self.lastname}, and the balance is {self.balance}."

guy = BankAccount("Chese", "Guy", "Baggins",20)

guy.deposit(10)
guy.withdrawl(50)
print(guy)

guy.deposit(100)
guy.withdrawl(50)
print(guy)

guy.deposit(1000)
guy.withdrawl(2)
print(guy)
```

```python
class Box: 
    def __init__(self, length, width):
        self.length = length
        self.width = width 
    def render(self): 
        for i in range(self.width):
            print("*" * self.length)
            
    def invert(self):
        x = self.width 
        y = self.length 
        self.width = y
        self.length = x 
        
    def get_area(self): 
        return ( self.width * self.length )
        
    def get_perimeter(self):
        return (2*(self.width + self.length))
    def double(self):
        x = self.width * 2 
        y = self.length * 2 

        self.width = x 
        self.length = y 
        return(self.width, self.length)
    def print_dim(self):
        print( f"The length is {self.length} and the width is {self.width}.")
    def get_dim(self):
        tup = (self.width, self.length)
        return tup
    def combine(self, other): 
        x = self.length + other.length
        y = self.width + other.width

        self.length = x 
        self.width = y 
    def get_hypot(self):
        hypot = ( self.length**2 + self.width**2 ) ** .5
        return hypot
    
    def __eq__(self,other):
        return self.width == other.width and self.length == other.length
```

```python
my_box = Box(4,5)

my_box.render()

my_box.invert()
print("\n")

my_box.render()
print("\n")

print(my_box.get_area())
print(my_box.get_perimeter())
```

```python
doubled = my_box.double()

print(doubled)

not_my_box = Box(2,3)

print(my_box == not_my_box)

my_box.print_dim()
print(my_box.get_dim())

my_box.combine(not_my_box)
my_box.print_dim()

print(my_box.get_hypot())

```

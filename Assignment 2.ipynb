```python
"""
Design a program that asks the user to enter a store’s sales for each day of the
week. The amounts should be stored in a list. Use a loop to calculate the total sales for
the week and display the result.
"""
Su = float(input('Enter sales for Sunday: '))
M  = float(input('Enter sales for Monday: '))
T  = float(input('Enter sales for Tuesday: '))
W  = float(input('Enter sales for Wednesday: '))
Th = float(input('Enter sales for Thursday: '))
F  = float(input('Enter sales for Friday: '))
Sa = float(input('Enter sales for Saturday: '))
```

```python
salesList = [ Su, M, T, W, Th, F, Sa ] 

x = 0

for i in salesList:
    x = i + x 
print( "The sales for the week are:", x)
```

```python
"""
Q3. Create a list with at least 5 places you’d like to travel to. Make sure the list isn’t in
alphabetical order
● Print your list in its original order.
● Use the sort() function to arrange your list in order and reprint your list.
● Use the sort(reverse=True) and reprint your list.
"""

travelList = [ 'Vancouver', 'The Tetons', 'Uganda', 'Hong Kong', 'Rome' ]

print(travelList)

travelList.sort()

print(travelList)

travelList.sort(reverse = True)

print(travelList)
```

```python
"""
Q5. Write a program that keeps names and email addresses in a dictionary as
key-value pairs. The program should then demonstrate the four options:
● look up a person’s email address,
● add a new name and email address,
● change an existing email address, and
● delete an existing name and email address.
"""

kRoe = { 
    "name":"Keenan",
    "email":"myemail@outlook.com",
}

dMiller = {
    "name":"Daniel",
    "email":"millertime@gmail.com"
}

jAnderson = {
    "name":"Jason",
    "email":"sonofander@yahoo.com"
}

usernameList = {"Keenan":kRoe, "Daniel":dMiller, "Jason":jAnderson
               }

print(usernameList)

def add_user(x,y): 
    new_user = {
        "name":x,
        "email":y
    }
    return new_user
```

```python
choice = input('would you like to look up the email of a user by name? Y/N') 

if choice == "Y":
    name = input('Enter the name you would like to look up: ') 
    print(usernameList[name]['email'])
```

```python
del choice 

choice = input('would you like to add a new user? Y/N')

if choice == 'Y':
    nm = input('input name:')
    em = input('input email:')

    usernameList[nm] = add_user(nm,em)
    
print(usernameList)
```

```python
del choice 

choice = input('would you like to change the email of an existing user?: Y/N') 

if choice == "Y":
    user = input('enter the user')
    newEmail = input('enter new email')
    
    usernameList[user]["email"] = newEmail

print(usernameList)
```

```python
del choice 

choice = input('would you like to delete a user? Y/N') 

if choice == "Y":
    user = input('What is the name of the user you would like to delete?') 

    del usernameList[user] 
print(usernameList)
```
